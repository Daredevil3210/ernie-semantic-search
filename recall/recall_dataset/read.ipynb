```python
with open('recall_result.txt', 'r', encoding='utf-8') as file:
    # 逐行读取前 10 行
    for i in range(50):
        line = file.readline()
        # 如果读到文件末尾，终止循环
        if not line:
            break
        print(line.strip())  # 打印每行，并移除行尾的换行符
```
