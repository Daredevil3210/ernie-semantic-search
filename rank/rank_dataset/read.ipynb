```python
with open('train_pairwise.csv', 'r', encoding='utf-8') as file:
    # 逐行读取前 10 行
    for i in range(50):
        line = file.readline()
        # 如果读到文件末尾，终止循环
        if not line:
            break
        print(line.strip())  # 打印每行，并移除行尾的换行符
```

```python
with open('train_pairwise.csv', 'r', encoding='utf-8') as f: #打开路径为data_path的文件
    flag = 0
    for line in f: #读取文件中的每一行
        if flag != 0 and flag < 10: #train_pairwise.csv中第一行不包含数据，因此要跳过
            data = line.rstrip().split("\t") #对读取的内容进行处理，rstrip()移除字符串末尾的空格或换行符，split("\t")按照Tab分割字符串，生成一个列表data。
            if len(data) != 3: #检查读取的这一行是否包含3个部分，确保数据的格式正确
                continue
            print(data)
        flag += 1
```
